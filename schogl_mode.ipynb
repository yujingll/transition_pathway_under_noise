```julia
import Pkg
Pkg.activate(".")
Pkg.instantiate()
using Peaks
using Plots, QuadGK, Dierckx, SpecialFunctions, LaTeXStrings
#multiplicative 
sigma = 0.3
U(x) = (k1*x^4)/4 - (k2*x^3)/3 + (k3*x^2)/2 - k4*x
k1 = 0.02;k2 = 0.20;k3 = 0.55;k4 = 0.4
f(x) = -k1*x^3+(k2*x^2)-k3*x+k4
g(x) = sigma*sqrt(k1*x^3 + k2*x^2 + k3*x + k4)
```

```julia
function export_reactive_density(sigma)
    g(x) = sigma
    x_grid = range(0,10,10^5);#grid for stationary distribution
    exp_int = zeros(100000) # ∫-U'[x]/g^2[x]
    Ps_grid = zeros(100000) # stationary distribution
    for i in 2:100000
        exp_int[i] = quadgk(x -> f(x)/g(x)^2, 0, x_grid[i], rtol=1e-5)[1]
        Ps_grid[i] = 1/g(x_grid[i])^2*exp(2*exp_int[i])
    end
    spl_Ps = Spline1D(x_grid, Ps_grid)
    Nps = integrate(spl_Ps, 0,10)#normalization constant 
    indices = findmaxima(Ps_grid/Nps)[1]
    left_min = x_grid[indices[1]] + 1; right_min = x_grid[indices[2]] - 1;
    q_range = range(left_min, right_min, 10^5)##grid for tpt
    
    exp_int_dq = zeros(100000) #∫U'[x]/g^2[x]
    dq_grid = zeros(100000) #dq = q'[x]
    for i in 2:100000
        exp_int_dq[i] = quadgk(x -> -f(x)/g(x)^2, 0, q_range[i], rtol=1e-5)[1]
        dq_grid[i] = exp(2*exp_int_dq[i])
    end
    #plot(q_range[2:end], dq_grid[2:end])#plot dq
    spl = Spline1D(q_range[2:end], dq_grid[2:end])
    Nq = integrate(spl, left_min, right_min) #normalization constant 
    q_grid = zeros(100000) # q_plus
    for i in 2:100000
        q_grid[i] = integrate(spl, left_min, q_range[i])
        q_grid[i]  = q_grid[i]/Nq
    end
    #plot(q_range, q_grid) # plot q_plus
    muAB_grid = zeros(100000)
    for i in 1:100000
        muAB_grid[i] = spl_Ps(q_range[i])/Nps * q_grid[i] * (1 - q_grid[i])
    end
    spl = Spline1D(q_range, muAB_grid)
    NmuAB = integrate(spl, left_min, right_min)#normalization constant 
    Vab, error =  quadgk(x -> spl(x)^2*g(x)^2*spl_Ps(x)*1/2/Nps, q_range[2], q_range[end], rtol=1e-5)
    return q_range, muAB_grid./NmuAB, Ps_grid/Nps, Vab, NmuAB/Vab
end
q_range, muAB_grid1_add, Ps_grid1 = export_reactive_density(0.2)
q_range, muAB_grid2_add,Ps_grid2 = export_reactive_density(0.4)
q_range, muAB_grid3_add,Ps_grid3 = export_reactive_density(0.6)
```

```julia
p3 = plot(range(0,10,10^5), Ps_grid1, label = L"\sigma = 0.2");
plot!(range(0,10,10^5), Ps_grid2, label = L"\sigma = 0.4", xlab = L"x",ylab = L"\rho(x)" ,legend =:topright,foreground_color_legend = nothing, grid = false, title = "Multiplicative noise")
plot!(range(0,10,10^5), Ps_grid3, label = L"\sigma = 0.6", xlab = L"x",ylab = L"\rho(x)" ,legend =:topright,foreground_color_legend = nothing, grid = false, title = "Multiplicative noise", labelsize = 1.5,  ytickfont=font(15), xtickfont=font(15), guidefont=font(25), legendfont = font(14), size = (600,400), framestyle = :box)
savefig("figure/Schlogl_model_add——density.pdf")
p3
```

```julia
function export_reactive_density_mul(sigma)
    g(x) = sigma*sqrt(k1*x^3 + k2*x^2 + k3*x + k4)
    x_grid = range(0,10,10^5);#grid for stationary distribution
    exp_int = zeros(100000) # ∫-U'[x]/g^2[x]
    Ps_grid = zeros(100000) # stationary distribution
    for i in 2:100000
        exp_int[i] = quadgk(x -> f(x)/g(x)^2, 0, x_grid[i], rtol=1e-5)[1]
        Ps_grid[i] = 1/g(x_grid[i])^2*exp(2*exp_int[i])
    end
    spl_Ps = Spline1D(x_grid, Ps_grid)
    Nps = integrate(spl_Ps, 0,10)#normalization constant 
    indices = findmaxima(Ps_grid/Nps)[1]
    left_min = x_grid[indices[1]] + 1; right_min = x_grid[indices[2]] - 1;
    q_range = range(left_min, right_min, 10^5)##grid for tpt
    
    exp_int_dq = zeros(100000) #∫U'[x]/g^2[x]
    dq_grid = zeros(100000) #dq = q'[x]
    for i in 2:100000
        exp_int_dq[i] = quadgk(x -> -f(x)/g(x)^2, 0, q_range[i], rtol=1e-5)[1]
        dq_grid[i] = exp(2*exp_int_dq[i])
    end
    #plot(q_range[2:end], dq_grid[2:end])#plot dq
    spl = Spline1D(q_range[2:end], dq_grid[2:end])
    Nq = integrate(spl, left_min, right_min) #normalization constant 
    q_grid = zeros(100000) # q_plus
    for i in 2:100000
        q_grid[i] = integrate(spl, left_min, q_range[i])
        q_grid[i]  = q_grid[i]/Nq
    end
    #plot(q_range, q_grid) # plot q_plus
    muAB_grid = zeros(100000)
    for i in 1:100000
        muAB_grid[i] = spl_Ps(q_range[i])/Nps * q_grid[i] * (1 - q_grid[i])
    end
    spl = Spline1D(q_range, muAB_grid)
    NmuAB = integrate(spl, left_min, right_min)#normalization constant 
    
    #JAB = 1/2*Ps_grid.*g.(x_grid).^2 .*dq_grid*1/Nps
    Vab, error =  quadgk(x -> spl(x)^2*g(x)^2*spl_Ps(x)*1/2/Nps, q_range[2], q_range[end], rtol=1e-5)
    return Ps_grid/Nps, q_range, muAB_grid./NmuAB, Vab, NmuAB/Vab
end

Ps_grid1, q_range1, muAB_grid1_mul, Vab1,Eab1  = export_reactive_density_mul(0.1)
Ps_grid2,q_range2, muAB_grid2_mul = export_reactive_density_mul(0.15)
Ps_grid3, q_range3, muAB_grid3_mul = export_reactive_density_mul(0.2)

```
